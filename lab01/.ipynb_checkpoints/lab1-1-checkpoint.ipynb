```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

# zad 1

```python
h= np.float64()
k = np.arange(0,17)
h = 10. ** (-k)
x = 1
eps = np.finfo(float).eps
exact = 1+np.tan(x)**2
```

```python
M = np.abs(2 * np.tan(x) * (1 + np.tan(x) ** 2))
method_error = M * h / 2
numeric_error = 2 * eps / h
computational_error = np.abs(exact-(np.tan(x + h) - np.tan(x)) / h)

h_min = h[np.argmin(computational_error)]
h_min_exact = 2 * np.sqrt(eps / M)
print(h_min,h_min_exact)
```

```python
plt.figure()
plt.loglog(h, method_error, label="method error")
plt.loglog(h, numeric_error, label="numeric error")
plt.loglog(h, computational_error, label="computational error")
plt.xlabel("h")
plt.ylabel("error")
plt.legend()
plt.show()
```

```python
M = np.abs(2 * (1 + np.tan(x) ** 2) * (1 + 3 * np.tan(x) ** 2))
method_error_central = M * h ** 2 / 6
numeric_error_central = eps / h
computational_error_central = np.abs(exact-(np.tan(x + h) - np.tan(x-h)) / (2*h))

h_min = h[np.argmin(computational_error_central)]
h_min_exact = (3 * eps / M) ** (1/3)
print(h_min,h_min_exact)
```

```python
plt.figure()
plt.loglog(h, method_error_central, label="method error")
plt.loglog(h, numeric_error_central, label="numeric error")
plt.loglog(h, computational_error_central, label="computational error")
plt.xlabel("h")
plt.ylabel("error")
plt.legend()
plt.show()
```

```python
forward = np.abs((np.tan(x+h) - np.tan(x))/h - exact)
central = np.abs((np.tan(x+h) - np.tan(x-h))/(2*h) - exact)

print("Forward:")
print("h_min emp =", h[np.argmin(forward)])
print("E_min     =", np.min(forward))

print("\nCentral:")
print("h_min emp =", h[np.argmin(central)])
print("E_min     =", np.min(central))

if np.min(central) < np.min(forward):
    print("\nDokładniejsza jest metoda różnic centralnych.")
else:
    print("\nDokładniejsza jest metoda różnic wprzód.")
```

# zad 2

```python
from decimal import Decimal, getcontext

getcontext().prec = 16

x = Decimal('1e-16')

original = (1 + x).sqrt() - 1
stable = x / ((1 + x).sqrt() + 1)

print("original:", original)
print("stable:", stable)
```

```python
y = np.exp(x)

f = (y - 1) / np.log(y)

print(f)
```

```python
plt.figure()
plt.semilogx(x,f)
plt.grid(True)
plt.title("f(x) = (y - 1) / log(y),  gdzie y = exp(x)", fontsize=12)
plt.show()
```

```python
ex = 1 + x/2 + x**2/6 + x**3/24
f = (ex - 1) / np.log(ex)

print(f)
```

```python
plt.figure()
plt.semilogx(x,f)
plt.grid(True)
plt.title("f(x) gdzie e^x ≈ 1 + x/2 + x²/6 + x³/24  – rozwinięcie Taylora", fontsize=12)
plt.show()
```

### liczenie za pomocą funkcji wbudowanej expm1(x)

```python
f = np.expm1(x)/x
print(f)
```

```python
plt.figure()
plt.semilogx(x,f)
plt.grid(True)
plt.title("f(x) = expm1(x) / x  – stabilna numerycznie metoda", fontsize=12)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.float64(10.0) ** (-np.arange(0, 17))

# 1. bezpośrednio
f1 = (np.exp(x) - 1) / x

# 2. z podstawieniem y = e^x
y = np.exp(x)
f2 = (y - 1) / np.log(y)

# 3. Taylor dla e^x
ex_taylor = 1 + x + x**2/2 + x**3/6 + x**4/24
f3 = (ex_taylor - 1) / np.log(ex_taylor)

# 4. stabilnie numerycznie
f4 = np.expm1(x) / x

# wspólny wykres
plt.figure(figsize=(10,6))
plt.semilogx(x, f1, 'o-', label='bezpośrednio: (exp(x)-1)/x')
plt.semilogx(x, f2, 's-', label='podstawienie: (y-1)/log(y)')
plt.semilogx(x, f3, '^-', label='Taylor')
plt.semilogx(x, f4, 'd-', label='expm1(x)/x')

plt.grid(True)
plt.legend()
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Porównanie metod obliczania f(x) = (exp(x)-1)/x')
plt.show()
```

Wszystkie metody prowadzą do wartości zbliżającej się do 1. Metoda Taylora jest mniej dokładna dla większych wartości x, ponieważ jest tylko przybliżeniem rozwinięcia funkcji. W metodzie z podstawieniem y = exp(x) oraz przy wykorzystaniu rozwinięcia Taylora pojawił się warning związany z dzieleniem przez bardzo małą wartość, co wynika z ograniczonej dokładności obliczeń zmiennoprzecinkowych. Warning nie wpływa jednak na końcowe wyniki, ponieważ wartości w tych punktach zostały poprawione. Najbardziej stabilne numerycznie wyniki daje metoda wykorzystująca funkcję expm1.

# zad 4

```python
n = [10 ** k for k in range(4,9)]
data = [np.random.uniform(0,1,10**k).astype(np.float32) for k in range(4,9)]
```

#### sumowanie w kolejności losowania. akumulator podwójnej precyzji

```python
def Sum_a(data):
    sum_A = np.float64(0)
    for x in data:
        sum_A += np.float64(x)
    return sum_A
```

#### sumowanie w kolejności losowania. akumulator pojedyńczej precyzji

```python
def Sum_b(data):
    sum_B = np.float32(0)
    for x in data:
        sum_B += np.float32(x)
    return sum_B
```

#### sumowanie za pomocą algorytmu Kahana


```python
def Sum_c(data):
    sum_C = np.float32(0)
    err = np.float32(0)
    for i in range(len(data)):
        y = np.float32(data[i] - err)
        temp = np.float32(sum_C + y)
        err = np.float32((temp - sum_C) - y)
        sum_C = temp
    return sum_C
```

#### sumowanie posortowanych danych rosnąco


```python
def Sum_d(data):
    data_D = np.copy(data)
    data_D.sort()
    sum_D = np.float32(0)
    for x in data_D:
        sum_D += np.float32(x)
    return sum_D
```

#### sumowanie posortowanych danych malejąco


```python
def Sum_e(data):
    data_E = np.copy(data)
    data_E.sort()
    data_E = data_E[::-1]
    sum_E = np.float32(0)
    for x in data_E:
        sum_E += np.float32(x)
    return sum_E
```

```python
def compute_relative_error(exact,s):
    return (abs(s - exact)) / abs(exact)
```

```python
results = []
relative_error = []
for i,d in enumerate(data):
    row = [math.fsum(d),Sum_a(d),Sum_b(d),Sum_c(d),Sum_d(d),Sum_e(d)]
    results.append(row)
    relative_error.append([compute_relative_error(row[0],row[j]) for j in range(len(row))])
```

```python
for res in relative_error:
    print(res)
```

```python
eps = 1e-20 

errors_plot = np.maximum(errors, eps)

plt.figure()

plt.plot(n, errors_plot[:,1], marker='o', label="float64")
plt.plot(n, errors_plot[:,2], marker='o', label="float32")
plt.plot(n, errors_plot[:,3], marker='o', label="Kahan")
plt.plot(n, errors_plot[:,4], marker='o', label="sort rosnąco")
plt.plot(n, errors_plot[:,5], marker='o', label="sort malejąco")

plt.xscale("log")
plt.yscale("log")

plt.xlabel("n")
plt.ylabel("błąd względny")
plt.title("Błąd względny metod sumowania")
plt.legend()
plt.grid(True)

plt.show()
```

Na wykresie przedstawiono zależność błędu względnego metod sumowania od liczby elementów n. Najmniejszy błąd uzyskano przy sumowaniu z akumulatorem podwójnej precyzji (float64), którego dokładność pozostaje bardzo wysoka dla wszystkich rozmiarów danych. Bardzo dobre wyniki daje również algorytm Kahana, którego błąd pozostaje niewielki i prawie stały. W przypadku sumowania w pojedynczej precyzji (float32) oraz sumowania posortowanych danych błąd rośnie wraz ze wzrostem liczby elementów, osiągając duże wartości dla największych zbiorów danych.
